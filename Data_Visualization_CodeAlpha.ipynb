```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv('https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv')

# Quick clean
df['Age'] = df['Age'].fillna(df['Age'].median())
df.drop(columns=['Cabin'], inplace=True)

print("Dataset ready! Shape:", df.shape)
```

```python
# Chart 1: Survival Count
plt.figure(figsize=(8,5))
sns.countplot(x='Survived', data=df, hue='Survived', palette='Set2', legend=False)
plt.title('Survival Count', fontsize=16, fontweight='bold')
plt.xlabel('Survived (0=Died, 1=Survived)', fontsize=12)
plt.ylabel('Number of Passengers', fontsize=12)
plt.xticks([0,1], ['Died', 'Survived'])
plt.savefig('chart1_survival_count.png', bbox_inches='tight')
plt.show()
print("Chart 1 saved!")
```

```python
# Chart 2: Survival by Gender
plt.figure(figsize=(8,5))
sns.countplot(x='Sex', data=df, hue='Survived', palette='Set1')
plt.title('Survival by Gender', fontsize=16, fontweight='bold')
plt.xlabel('Gender', fontsize=12)
plt.ylabel('Number of Passengers', fontsize=12)
plt.legend(['Died', 'Survived'])
plt.savefig('chart2_survival_gender.png', bbox_inches='tight')
plt.show()
print("Chart 2 saved!")
```

```python
# Chart 3: Survival by Passenger Class
plt.figure(figsize=(8,5))
sns.countplot(x='Pclass', data=df, hue='Survived', palette='Set2')
plt.title('Survival by Passenger Class', fontsize=16, fontweight='bold')
plt.xlabel('Passenger Class (1=First, 2=Second, 3=Third)', fontsize=12)
plt.ylabel('Number of Passengers', fontsize=12)
plt.legend(['Died', 'Survived'])
plt.savefig('chart3_survival_class.png', bbox_inches='tight')
plt.show()
print("Chart 3 saved!")
```

```python
# Chart 4: Age Distribution by Survival
plt.figure(figsize=(10,5))
sns.histplot(data=df, x='Age', hue='Survived', bins=30, kde=True, palette='Set1')
plt.title('Age Distribution by Survival', fontsize=16, fontweight='bold')
plt.xlabel('Age', fontsize=12)
plt.ylabel('Number of Passengers', fontsize=12)
plt.legend(['Survived', 'Died'])
plt.savefig('chart4_age_distribution.png', bbox_inches='tight')
plt.show()
print("Chart 4 saved!")
```

```python
# Chart 5: Survival Percentage Pie Chart
survival_counts = df['Survived'].value_counts()
plt.figure(figsize=(7,7))
plt.pie(survival_counts,
        labels=['Died', 'Survived'],
        autopct='%1.1f%%',
        colors=['#FF6B6B', '#51CF66'],
        startangle=90,
        explode=(0.05, 0.05))
plt.title('Survival Percentage', fontsize=16, fontweight='bold')
plt.savefig('chart5_survival_pie.png', bbox_inches='tight')
plt.show()
print("Chart 5 saved!")
```

```python
# Chart 6: Correlation Heatmap
plt.figure(figsize=(10,8))
numeric_df = df.select_dtypes(include=['number'])
correlation = numeric_df.corr()
sns.heatmap(correlation,
            annot=True,
            fmt='.2f',
            cmap='coolwarm',
            linewidths=0.5)
plt.title('Correlation Heatmap', fontsize=16, fontweight='bold')
plt.savefig('chart6_heatmap.png', bbox_inches='tight')
plt.show()
print("Chart 6 saved!")
```

```python
# Chart 7: Dashboard - All charts in one figure
fig, axes = plt.subplots(2, 2, figsize=(14,10))
fig.suptitle('Titanic Data Analysis Dashboard', fontsize=20, fontweight='bold')

# Chart 1
sns.countplot(x='Survived', data=df, hue='Survived', palette='Set2', legend=False, ax=axes[0,0])
axes[0,0].set_title('Survival Count')
axes[0,0].set_xticks([0,1])
axes[0,0].set_xticklabels(['Died', 'Survived'])

# Chart 2
sns.countplot(x='Sex', data=df, hue='Survived', palette='Set1', ax=axes[0,1])
axes[0,1].set_title('Survival by Gender')
axes[0,1].legend(['Died', 'Survived'])

# Chart 3
sns.countplot(x='Pclass', data=df, hue='Survived', palette='Set2', ax=axes[1,0])
axes[1,0].set_title('Survival by Class')
axes[1,0].legend(['Died', 'Survived'])

# Chart 4
sns.histplot(data=df, x='Age', hue='Survived', bins=30, palette='Set1', ax=axes[1,1])
axes[1,1].set_title('Age Distribution')

plt.tight_layout()
plt.savefig('chart7_dashboard.png', bbox_inches='tight')
plt.show()
print("Dashboard saved!")
```

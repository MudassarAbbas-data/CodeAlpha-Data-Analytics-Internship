```python
# Loading Titanic dataset directly from internet
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv')

# Show first 5 rows
df.head()
```

```python
# How many rows and columns?
print("Shape of dataset:", df.shape)
```

```python
# What columns do we have?
print("Columns:", df.columns.tolist())
```

```python
# What type of data is in each column?
print(df.dtypes)
```

```python
# How many missing values in each column?
print(df.isnull().sum())
```

```python
# Fill missing Age with median age
df['Age'] = df['Age'].fillna(df['Age'].median())

# Fill missing Embarked with most common value
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# Verify no more missing values
print(df.isnull().sum())
```

```python
# Statistical summary of the data
df.describe()
```

```python
# How many survived vs died?
plt.figure(figsize=(8,5))
sns.countplot(x='Survived', data=df, hue='Survived', palette='Set2', legend=False)
plt.title('Survival Count (0=Died, 1=Survived)')
plt.xlabel('Survived')
plt.ylabel('Number of Passengers')
plt.show()
```

```python
# Did gender affect survival?
plt.figure(figsize=(8,5))
sns.countplot(x='Sex', data=df, hue='Survived', palette='Set1')
plt.title('Survival by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Passengers')
plt.legend(['Died', 'Survived'])
plt.show()
```

```python
# Did ticket class affect survival?
plt.figure(figsize=(8,5))
sns.countplot(x='Pclass', data=df, hue='Survived', palette='Set2')
plt.title('Survival by Passenger Class')
plt.xlabel('Passenger Class (1=First, 2=Second, 3=Third)')
plt.ylabel('Number of Passengers')
plt.legend(['Died', 'Survived'])
plt.show()
```

```python
# How were passengers distributed by age?
plt.figure(figsize=(8,5))
sns.histplot(df['Age'], bins=30, kde=True, color='steelblue')
plt.title('Age Distribution of Passengers')
plt.xlabel('Age')
plt.ylabel('Number of Passengers')
plt.show()
```

```python
# Did paying more fare increase survival?
plt.figure(figsize=(8,5))
sns.boxplot(x='Survived', y='Fare', data=df, palette='Set3')
plt.title('Fare vs Survival')
plt.xlabel('Survived (0=Died, 1=Survived)')
plt.ylabel('Fare Paid')
plt.show()
```
